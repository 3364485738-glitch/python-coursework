## 一、字符串的乐趣

本部分实现用于处理字符串的简单函数。按题目要求，这些函数不需要执行错误检查。

### 1. 判断回文

回文指从前往后读和从后往前读都一样的词或短语。判断时应忽略空格和大小写。

```python
def is_palindrome(text):
    # 只保留字母和数字，并转换为小写
    cleaned = [ch for ch in text if ch not in " "]
    cleaned = [ch.lower() for ch in cleaned]

    return cleaned == cleaned[::-1]


# 测试
test_cases = [
    "level",
    "kayak",
    "pop",
    "tacocat",
    "T A C O cat",
    "rats live on no evil star",
    "Was it a car or a cat I saw?",
    "python",
]

for text in test_cases:
    print(f"{text!r} 是否为回文：{is_palindrome(text)}")
```

先把输入中的空格去掉并统一转换为小写，再判断清理后的序列与它的逆序是否相同。tacocat 和 T A C O cat 清理后都是 tacocat，因此都被认为是回文。

### 2. 判断字母顺序的词

如果词的字母从左到右按字母顺序排列（允许重复），就称它是字母顺序的。判断时应忽略空格。

```python
def is_abecedarian(word):
    # 去掉空格
    letters = [ch for ch in word if ch != " "]

    # 相邻字母必须按字母顺序排列（允许相等）
    for i in range(len(letters) - 1):
        if letters[i] > letters[i + 1]:
            return False

    return True


# 测试
test_cases = ["adder", "beet", "dog", "cat", "abcd efgh xyz", "almost"]

for word in test_cases:
    print(f"{word!r} 是否为字母顺序：{is_abecedarian(word)}")
```

逐对比较相邻字母，只要出现前一个字母大于后一个字母的情况就不是字母顺序的。abcd efgh xyz 忽略空格后字母依次递增，因此被认为是字母顺序的。

### 3. 去掉所有元音

```python
def remove_vowels(text):
    vowels = "aeiouAEIOU"

    return "".join(ch for ch in text if ch not in vowels)


# 测试
print("remove_vowels('cat') =", remove_vowels("cat"))
print("remove_vowels('audio') =", remove_vowels("audio"))
print("remove_vowels('aeiou') =", remove_vowels("Aeiou"))
print("remove_vowels('rhythm') =", remove_vowels("rhythm"))
```

使用生成器表达式逐个保留非元音字符，再用 join 连接成新字符串。全部由元音组成的字符串（如 audio 的元音部分、纯元音字符串）经过处理后得到空字符串，函数也能正确处理。

## 二、列表的乐趣

### 1. 反转列表

list_reverse 接受一个列表，返回反转后的列表；输入不是列表时引发适当的错误。

```python
def list_reverse(items):
    if not isinstance(items, list):
        raise TypeError("输入必须是列表")

    return items[::-1]


# 测试
print("list_reverse([1, 2, 3]) =", list_reverse([1, 2, 3]))
print("list_reverse(['a', 'b', 'c']) =", list_reverse(["a", "b", "c"]))
print("list_reverse([]) =", list_reverse([]))

# 输入不是列表时应引发错误
for bad_input in ["abc", (1, 2, 3)]:
    try:
        list_reverse(bad_input)
    except TypeError as error:
        print(f"输入 {bad_input!r} 引发错误：{error}")
```

使用切片 [::-1] 生成反转后的新列表，原列表保持不变。isinstance 检查输入类型，不是列表时引发 TypeError。

### 2. 判断序列是否按升序排列

is_sorted 只遍历序列一次；seq 是支持索引和比较的 Python 序列，无需检查输入类型。

```python
def is_sorted(seq):
    # 只遍历一次：逐对比较相邻元素
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False

    return True


# 测试
test_cases = [
    [1, 2, 3, 4],
    [1, 3, 2],
    ("a", "b", "c"),
    "abc",
    [5],
    [],
    [2, 2, 2],
]

for seq in test_cases:
    print(f"is_sorted({seq!r}) = {is_sorted(seq)}")
```

函数只遍历序列一次：逐对比较相邻元素，一旦出现逆序立即返回 False。只要 seq 支持索引、元素支持比较，函数就能工作，列表、元组、字符串都能正确判断，体现了 Python 动态类型的特点。

## 三、作为向量的元组

用长度为 d 的元组表示 d 维向量。

### 1. 向量与标量相乘

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")

    if not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("元组中的元素必须是整数或浮点数")

    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或浮点数")

    return tuple(x * s for x in t)


# 测试
print("vec_scalar_mult((1, 2, 3), 2) =", vec_scalar_mult((1, 2, 3), 2))
print("vec_scalar_mult((1.5, -2), 0.5) =", vec_scalar_mult((1.5, -2), 0.5))
print("vec_scalar_mult((4, 5), -1) =", vec_scalar_mult((4, 5), -1))

# 错误输入测试
for bad_t, bad_s in [([1, 2], 2), ((1, "a"), 2), ((1, 2), "x")]:
    try:
        vec_scalar_mult(bad_t, bad_s)
    except TypeError as error:
        print(f"输入 ({bad_t!r}, {bad_s!r}) 引发错误：{error}")
```

函数分别检查元组本身、元组元素和标量的类型，类型不合适时引发带有适当错误消息的 TypeError。整数标量和含整数元素的元组都是有效输入，结果与浮点输入格式一致。

### 2. 向量内积

向量 x, y ∈ R^d 的内积是 Σ xⱼyⱼ；任一输入不是元组时引发 TypeError，长度不一致时引发 ValueError。

```python
def vec_inner_product(u, v):
    if not isinstance(u, tuple) or not isinstance(v, tuple):
        raise TypeError("两个输入都必须是元组")

    if len(u) != len(v):
        raise ValueError("两个元组的长度必须一致")

    return float(sum(x * y for x, y in zip(u, v)))


# 测试
print("vec_inner_product((1, 2, 3), (4, 5, 6)) =", vec_inner_product((1, 2, 3), (4, 5, 6)))
print("vec_inner_product((1.5, 2), (2, 1)) =", vec_inner_product((1.5, 2), (2, 1)))
print("vec_inner_product((1, 2), (3, 4, 5)) 的长度问题见下：")

# 错误输入测试
try:
    vec_inner_product([1, 2], (3, 4))
except TypeError as error:
    print("输入不是元组：", error)

try:
    vec_inner_product((1, 2), (3, 4, 5))
except ValueError as error:
    print("长度不一致：", error)
```

用 zip 将两个元组按位置配对，累加对应元素乘积并以 float 返回。输入类型错误引发 TypeError，长度不一致引发 ValueError。

### 3. 用元组的元组表示矩阵与有效性检查

外层元组长度为 m，每个内层元组长度为 n，表示矩阵的一行。

```python
def check_valid_mx(matrix):
    # 外层必须是元组，且每个元素本身都是元组
    if not isinstance(matrix, tuple):
        return False

    if not all(isinstance(row, tuple) for row in matrix):
        return False

    # 所有内层元组长度相同
    if len(set(len(row) for row in matrix)) > 1:
        return False

    # 每个内层元组的每个元素都是整数或浮点数
    for row in matrix:
        for value in row:
            if not isinstance(value, (int, float)) or isinstance(value, bool):
                return False

    return True


# 测试
my_mx = ((1, 2, 3), (4, 5, 6))

print("check_valid_mx(my_mx) =", check_valid_mx(my_mx))
print("check_valid_mx(((1, 2), (3, 4, 5))) =", check_valid_mx(((1, 2), (3, 4, 5))))
print("check_valid_mx(((1, 'a'), (2, 3))) =", check_valid_mx(((1, "a"), (2, 3))))
print("check_valid_mx([1, 2, 3]) =", check_valid_mx([1, 2, 3]))
print("check_valid_mx(((1, [2]), (3, 4))) =", check_valid_mx(((1, [2]), (3, 4))))
print("check_valid_mx(()) =", check_valid_mx(()))
```

按三条规则依次检查：外层元组的每个元素都是元组、所有内层元组长度相同、每个元素都是整数或浮点数。isinstance(value, bool) 的排除是必要的，因为 Python 中 bool 是 int 的子类，True 会被误判为整数。

### 4. 矩阵与向量相乘

```python
def mx_vec_mult(matrix, vector):
    if not check_valid_mx(matrix):
        raise TypeError("矩阵必须是由整数或浮点数组成的元组的元组")

    if not isinstance(vector, tuple):
        raise TypeError("向量必须是元组")

    if not all(isinstance(x, (int, float)) for x in vector):
        raise TypeError("向量中的元素必须是整数或浮点数")

    row_count = len(matrix)
    column_count = len(matrix[0]) if row_count > 0 else 0

    if len(vector) != column_count:
        raise ValueError("向量长度必须等于矩阵的列数")

    return tuple(
        float(sum(value * x for value, x in zip(row, vector)))
        for row in matrix
    )


# 测试
my_mx = ((1, 2, 3), (4, 5, 6))
vec = (1, 0, -1)

print("矩阵：", my_mx)
print("向量：", vec)
print("mx_vec_mult(my_mx, vec) =", mx_vec_mult(my_mx, vec))

# 错误输入测试
try:
    mx_vec_mult(((1, 2), (3,)), vec)
except TypeError as error:
    print("矩阵不合法：", error)

try:
    mx_vec_mult(my_mx, (1, 2))
except ValueError as error:
    print("维度不匹配：", error)
```

先复用 check_valid_mx 检查矩阵，再检查向量类型和维度匹配。结果是与矩阵行数相同的数值元组，每个元素是对应行与向量的内积。例如第一行 (1, 2, 3)·(1, 0, -1) = 1×1 + 2×0 + 3×(−1) = −2。

## 四、更多向量的乐趣

### 1. 定义 Vector 类

每个向量有维度（非负整数）和元素列表或元组。只提供维度时，默认元素为全 0；维度类型不正确或为负数、维度与元素个数不一致时抛出合适的异常。

```python
class Vector:
    def __init__(self, dimension, elements=None):
        # 维度必须是非负整数
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("维度必须是整数")

        if dimension < 0:
            raise ValueError("维度不能为负数")

        # 只提供维度时，默认使用全 0 元素
        if elements is None:
            elements = [0] * dimension

        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")

        if len(elements) != dimension:
            raise ValueError("元素个数必须与维度一致")

        self.dimension = dimension
        self.elements = list(elements)

    def __repr__(self):
        return f"Vector({self.dimension}, {tuple(self.elements)})"


# 测试
v1 = Vector(3, (1, 2, 3))
v2 = Vector(2, [4, 5])
v3 = Vector(3)  # 只提供维度

print("v1 =", v1)
print("v2 =", v2)
print("v3 =", v3)

# 异常测试
for bad_args in [("three", (1, 2)), (-1, (1,)), (3, (1, 2))]:
    try:
        Vector(*bad_args)
    except (TypeError, ValueError) as error:
        print(f"Vector{bad_args} 引发异常：{type(error).__name__}: {error}")
```

### 2. 维度和元素应是类属性还是实例属性？

**应选择实例属性。**理由如下：

- 每个向量都有自己的维度和元素，不同 Vector 对象之间这些值各不相同；如果把它们定义为类属性，所有实例将共享同一份数据，修改任何一个向量的元素都会影响全部向量，这显然不符合向量的语义。
- 类属性适合描述所有实例共享的常量信息，而维度和元素是每个对象在初始化时确定的个体状态，因此必须在 __init__ 中通过 self 赋值为实例属性。

### 3. 实现比较运算符

坐标相同则相等；否则按 Python 元组的字典序比较。比较维度不同的向量时抛出异常。

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("维度必须是整数")

        if dimension < 0:
            raise ValueError("维度不能为负数")

        if elements is None:
            elements = [0] * dimension

        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")

        if len(elements) != dimension:
            raise ValueError("元素个数必须与维度一致")

        self.dimension = dimension
        self.elements = list(elements)

    # ---- 辅助方法：检查维度一致 ----
    def _check_same_dimension(self, other):
        if not isinstance(other, Vector):
            raise TypeError("比较对象必须是 Vector")

        if self.dimension != other.dimension:
            raise ValueError("两个向量的维度必须一致")

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented

        return self.elements == other.elements

    def __lt__(self, other):
        self._check_same_dimension(other)

        return tuple(self.elements) < tuple(other.elements)

    def __le__(self, other):
        self._check_same_dimension(other)

        return tuple(self.elements) <= tuple(other.elements)

    def __gt__(self, other):
        self._check_same_dimension(other)

        return tuple(self.elements) > tuple(other.elements)

    def __ge__(self, other):
        self._check_same_dimension(other)

        return tuple(self.elements) >= tuple(other.elements)

    def __repr__(self):
        return f"Vector({self.dimension}, {tuple(self.elements)})"


# 测试
a = Vector(2, (2, 4))
b = Vector(2, (2, 5))
c = Vector(2, (2, 4))
d = Vector(3, (1, 2, 3))

print("a =", a, "，b =", b, "，c =", c)
print("a == c：", a == c)
print("a == b：", a == b)
print("a < b：", a < b)
print("a <= c：", a <= c)
print("b > a：", b > a)
print("a >= b：", a >= b)

# 维度不同时抛出异常
try:
    a < d
except ValueError as error:
    print("维度不同的向量比较：", error)

# 与非 Vector 对象比较
print("a == (2, 4)：", a == (2, 4))
```

相等判断只要求坐标完全相同；小于、小于等于、大于等运算符把元素转成元组后利用 Python 元组自带的字典序进行比较。维度不同时抛出 ValueError。与非 Vector 对象比较相等时返回 NotImplemented，由 Python 回退为默认的不相等结果，而不是报错。

### 4. 实现 Vector.dot 方法

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("维度必须是整数")

        if dimension < 0:
            raise ValueError("维度不能为负数")

        if elements is None:
            elements = [0] * dimension

        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")

        if len(elements) != dimension:
            raise ValueError("元素个数必须与维度一致")

        self.dimension = dimension
        self.elements = list(elements)

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot 的参数必须是 Vector 对象")

        if self.dimension != other.dimension:
            raise ValueError("两个向量的维度必须一致")

        return float(sum(x * y for x, y in zip(self.elements, other.elements)))

    def __repr__(self):
        return f"Vector({self.dimension}, {tuple(self.elements)})"


v = Vector(3, (1, 2, 3))
w = Vector(3, (4, 5, 6))

print("v =", v, "，w =", w)
print("v.dot(w) =", v.dot(w))

# 异常测试
try:
    v.dot((4, 5, 6))
except TypeError as error:
    print("参数不是 Vector：", error)

try:
    v.dot(Vector(2, (1, 2)))
except ValueError as error:
    print("维度不一致：", error)
```

dot 方法先检查参数类型和维度，再把两个向量的对应元素相乘并求和，以浮点数形式返回内积。v·w = 1×4 + 2×5 + 3×6 = 32。

### 5. 实现乘法运算符

标量左乘或右乘返回各元素都乘以该标量的新 Vector；两个 Vector 相乘时对应位置元素逐位相乘，结果仍是新 Vector。

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("维度必须是整数")

        if dimension < 0:
            raise ValueError("维度不能为负数")

        if elements is None:
            elements = [0] * dimension

        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")

        if len(elements) != dimension:
            raise ValueError("元素个数必须与维度一致")

        self.dimension = dimension
        self.elements = list(elements)

    def __mul__(self, other):
        # 右乘：v * 标量 或 v * w
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return Vector(self.dimension, [x * other for x in self.elements])

        if isinstance(other, Vector):
            if self.dimension != other.dimension:
                raise ValueError("两个向量的维度必须一致")

            return Vector(self.dimension, [x * y for x, y in zip(self.elements, other.elements)])

        raise TypeError("乘法对象必须是标量或 Vector 对象")

    def __rmul__(self, other):
        # 左乘：标量 * v，转交给 __mul__ 处理
        return self.__mul__(other)

    def __repr__(self):
        return f"Vector({self.dimension}, {tuple(self.elements)})"


v = Vector(2, (2, 3))
w = Vector(2, (4, 5))

print("v =", v, "，w =", w)
print("2 * v =", 2 * v)
print("v * 2 =", v * 2)
print("v * w =", v * w)

# 异常测试
try:
    v * Vector(3, (1, 2, 3))
except ValueError as error:
    print("维度不一致：", error)

try:
    v * "abc"
except TypeError as error:
    print("非法乘法对象：", error)
```

__mul__ 同时处理标量乘法和逐位乘法：乘数是数字时返回各元素乘以该标量的新 Vector，是 Vector 时逐位相乘并要求维度一致。__rmul__ 处理标量在左边的情形（2 * v），当 Python 发现整数不支持与 Vector 相乘时，会自动尝试 v 的 __rmul__，因此 2 * v 和 v * 2 的结果一致。所有运算都返回新对象，原向量不被修改。
