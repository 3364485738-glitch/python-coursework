### 实验2：函数、条件、循环与递归

#### 1. 成绩等级判断
根据成绩所在区间返回等级，超出0～100范围时抛出ValueError。

```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在0～100之间")

    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


# 测试各区间及边界
for score in [0, 59, 60, 79, 80, 89, 90, 100]:
    print(f"{score}分：{score_level(score)}")

# 测试超出范围的成绩
for score in [-1, 101]:
    try:
        print(score_level(score))
    except ValueError as error:
        print(f"{score}分：{error}")
```

#### 2. 使用for和while求和
分别使用两种循环计算1到100的整数和，并比较结果。

```python
# for 循环
total_for = 0

for number in range(1, 101):
    total_for += number

# while 循环
total_while = 0
number = 1

while number <= 100:
    total_while += number
    number += 1

print("for 循环结果：", total_for)
print("while 循环结果：", total_while)
print("两种结果是否相等：", total_for == total_while)
```

range(1, 101)生成1到100的整数，不包含结束值101。
for循环依次遍历这些整数；while循环在条件成立时重复执行，并手动更新number。
两种方法的计算结果相同，均为5050。

#### 3. 递归与迭代计算阶乘
编写两种阶乘函数，输入为非负整数，约定0! = 1。

```python
# 递归版：函数调用自己
def factorial_recursive(n):
    if type(n) is not int or n < 0:
        raise ValueError("n必须是非负整数")

    if n == 0:
        return 1

    return n * factorial_recursive(n - 1)


# 迭代版：使用循环累乘
def factorial_iterative(n):
    if type(n) is not int or n < 0:
        raise ValueError("n必须是非负整数")

    result = 1

    for number in range(1, n + 1):
        result *= number

    return result


# 比较题目要求的四组输入
for n in [0, 1, 5, 10]:
    recursive_result = factorial_recursive(n)
    iterative_result = factorial_iterative(n)

    print(
        f"n={n}，递归结果={recursive_result}，"
        f"迭代结果={iterative_result}，"
        f"是否相等={recursive_result == iterative_result}"
    )
```

递归版利用n! = n × (n−1)!，并以0! = 1作为终止条件。
迭代版从1开始，通过循环不断累乘。
当n为0时，迭代版循环不执行，返回初始值1。
四组输入下两种方法结果一致。递归版受递归深度限制，较大输入更适合使用迭代版。

#### 4. 统计函数与元组解包
函数接收非空数值列表，以元组返回最小值、最大值和平均值。

```python
def summary(values):
    if len(values) == 0:
        raise ValueError("列表不能为空")

    minimum = min(values)
    maximum = max(values)
    average = sum(values) / len(values)

    return minimum, maximum, average


values = [85, 58, 92, 76, 60, 89]

# 接收完整的元组
result = summary(values)
print("返回结果：", result)
print("返回类型：", type(result))

# 元组解包
minimum, maximum, average = result

print("最小值：", minimum)
print("最大值：", maximum)
print(f"平均值：{average:.2f}")
```

min()、max()分别计算最小值和最大值，sum(values) / len(values)计算平均值。
return minimum, maximum, average将三个值组成元组返回。
元组解包按顺序将三个结果赋给对应变量。

#### 5. 使用可变数量参数求和
定义sum_all(*args)，分别测试无参数、单个参数和多个参数。

```python
def sum_all(*args):
    return sum(args)


print("无参数：", sum_all())
print("单个参数：", sum_all(10))
print("多个参数：", sum_all(1, 2, 3, 4, 5))
```

*args接收数量不固定的位置参数，并将它们保存为元组。
sum(args)计算所有参数之和。
无参数时args为空元组，sum()返回0；单个参数时返回该数值；多个参数时返回它们的总和。

```python
data_dir = "exp01_data"
students_path = os.path.join(data_dir, "students.txt")

os.makedirs(data_dir, exist_ok=True)

print("数据目录：", os.path.abspath(data_dir))
print("学生文件路径：", students_path)
```

#### 2. 写入并读取UTF-8文本文件
使用with open管理文件，写入模拟成绩数据，再读取并跳过表头。

```python
students_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

# 写入文件
with open(students_path, "w", encoding="utf-8") as file:
    file.write(students_text)

# 读取文件
with open(students_path, "r", encoding="utf-8") as file:
    header = next(file)
    student_lines = file.readlines()

print("表头：", header.strip())
print("数据行数：", len(student_lines))

# 使用字符串方法拆分字段
for line in student_lines:
    fields = line.strip().split(",")
    print(fields)
```

#### 3. 定义parse_student函数
检查字段数量、姓名、成绩格式和成绩范围，无效记录抛出ValueError。

```python
def parse_student(line):
    fields = line.strip().split(",")

    if len(fields) != 2:
        raise ValueError("字段数量必须为2")

    name = fields[0].strip()
    score_text = fields[1].strip()

    if name == "":
        raise ValueError("姓名不能为空")

    try:
        score = int(score_text)
    except ValueError:
        raise ValueError("成绩必须能转换为整数") from None

    if not 0 <= score <= 100:
        raise ValueError("成绩必须在0～100之间")

    return {"name": name, "score": score}


print("有效记录：", parse_student("Alice,85"))
```

#### 4. 逐行解析并处理异常
有效记录保存为字典组成的列表，无效记录保存行号与错误原因，并继续处理后续记录。

```python
valid_students = []
invalid_records = []

with open(students_path, "r", encoding="utf-8") as file:
    next(file)  # 跳过第1行表头

    for line_number, line in enumerate(file, start=2):
        try:
            student = parse_student(line)
            valid_students.append(student)
        except ValueError as error:
            invalid_records.append({
                "line_number": line_number,
                "reason": str(error),
                "content": line.strip()
            })

print("有效记录：")
for student in valid_students:
    print(student)

print("\n无效记录：")
for record in invalid_records:
    print(
        f"第{record['line_number']}行，"
        f"内容：{record['content']}，"
        f"原因：{record['reason']}"
    )

print("\n有效记录数：", len(valid_students))
print("无效记录数：", len(invalid_records))
```

#### 5. 保存统计报告并比较写入模式
w模式覆盖已有内容，a模式在已有内容末尾追加。

```python
valid_count = len(valid_students)
invalid_count = len(invalid_records)

valid_scores = [student["score"] for student in valid_students]
average_score = sum(valid_scores) / valid_count
passed_count = sum(score >= 60 for score in valid_scores)
pass_rate = passed_count / valid_count

report_path = os.path.join(data_dir, "report.txt")

# 先写入一段内容，用于观察覆盖效果
with open(report_path, "w", encoding="utf-8") as file:
    file.write("这是一段将被覆盖的旧内容。\n")

# 再次使用 w，旧内容被覆盖
with open(report_path, "w", encoding="utf-8") as file:
    file.write("学生成绩统计报告\n")
    file.write(f"有效记录数：{valid_count}\n")
    file.write(f"无效记录数：{invalid_count}\n")
    file.write(f"平均分：{average_score:.2f}\n")
    file.write(f"及格率：{pass_rate:.2%}\n")

print("使用 w 后：")
with open(report_path, "r", encoding="utf-8") as file:
    print(file.read())

# 使用 a，在末尾追加
with open(report_path, "a", encoding="utf-8") as file:
    file.write("备注：无效记录未参与平均分及及格率计算。\n")

print("使用 a 后：")
with open(report_path, "r", encoding="utf-8") as file:
    print(file.read())
```

w模式在打开已有文件时清空原内容，再写入新内容；文件不存在时会创建文件。
a模式保留已有内容，将新内容追加到末尾；文件不存在时也会创建文件。

### 实验5：类、继承与对象复制

#### 1. 定义Student类
初始化姓名和成绩属性，实现及格判断和对象信息输出。

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"姓名：{self.name}，成绩：{self.score}"


# 创建一个对象并测试
example_student = Student("Alice", 85)

print(example_student)
print("是否及格：", example_student.is_passed())
```

#### 2. 将有效记录转换为Student对象
使用实验4中的有效记录，逐个输出学生信息及及格状态。

```python
# valid_students 来自实验4
student_objects = [
    Student(record["name"], record["score"])
    for record in valid_students
]

for student in student_objects:
    status = "及格" if student.is_passed() else "不及格"
    print(f"{student}，状态：{status}")

print("对象数量：", len(student_objects))
```

#### 3. 定义GraduateStudent子类
继承Student，通过super()初始化父类属性，增加研究主题，并重写字符串表示方法。

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (
            f"姓名：{self.name}，成绩：{self.score}，"
            f"研究主题：{self.research_topic}"
        )


graduate_student = GraduateStudent("Carol", 92, "数据分析")

print(graduate_student)
print("是否及格：", graduate_student.is_passed())
print("是否属于Student类：", isinstance(graduate_student, Student))
```

GraduateStudent继承Student，因此可以使用父类的is_passed()方法。
super().__init__()调用父类的初始化方法，设置姓名和成绩。
子类增加research_topic属性，并重写__str__()，使输出包含研究主题。

#### 4. 比较直接赋值、浅拷贝与深拷贝
分别修改嵌套列表，观察原对象和复制对象的变化。每次使用新建的原始对象。

```python
import copy

# 第一组：直接赋值
original = {"name": "Alice", "scores": [85, 90]}
assigned = original

assigned["scores"][0] = 100

print("直接赋值：")
print("原对象：", original)
print("赋值对象：", assigned)
print("外层对象是否相同：", original is assigned)


# 第二组：浅拷贝，重新创建原始对象
original = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original)

shallow["scores"][0] = 100

print("\n浅拷贝：")
print("原对象：", original)
print("浅拷贝对象：", shallow)
print("外层对象是否相同：", original is shallow)
print("内部列表是否相同：", original["scores"] is shallow["scores"])


# 第三组：深拷贝，再次重新创建原始对象
original = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original)

deep["scores"][0] = 100

print("\n深拷贝：")
print("原对象：", original)
print("深拷贝对象：", deep)
print("外层对象是否相同：", original is deep)
print("内部列表是否相同：", original["scores"] is deep["scores"])
```

直接赋值不创建新对象，两个变量指向同一个字典。
浅拷贝创建新的外层字典，但内部列表仍然共享，所以修改列表元素也会改变原对象。
深拷贝同时复制内部列表，本例修改深拷贝对象不会影响原对象。
